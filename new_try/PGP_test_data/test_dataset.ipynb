````python
import os
import json
import time
from pathlib import Path

import numpy as np
import pandas as pd
from dotenv import load_dotenv
from anthropic import Anthropic
````

````python
# ========================================
# 1. API 설정
# ========================================

load_dotenv(override=True)

API_KEY = os.getenv("CLAUDE_API_KEY")
MODEL = os.getenv("CLAUDE_DEFAULT_MODEL", "claude-haiku-5-5")

if not API_KEY:
    raise RuntimeError(".env에 CLAUDE_API_KEY를 설정하세요.")

client = Anthropic(api_key=API_KEY, timeout=120.0, max_retries=2)

print("사용 모델:", MODEL)
````

````python
# ========================================
# 2. 데이터 불러오기
# ========================================

health_data = pd.read_excel("test_health_data.xlsx")

DISEASES = ["당뇨", "심장", "뇌혈관", "신장", "간", "고혈압"]

BASIC_COLS = ["participant", "성별", "나이"]

DIAGNOSIS_MAP = {
    "당뇨": ["당뇨"],
    "심장": [
        "관상동맥질환", "협심증", "심근경색",
        "심부전", "심장질환_허혈성또는심부전"
    ],
    "뇌혈관": ["뇌졸중"],
    "신장": [],
    "간": [],
    "고혈압": ["고혈압"]
}

DIAGNOSIS_COLS = [
    "당뇨", "당뇨_1형", "당뇨_2형", "당뇨_유형미상",
    "뇌졸중", "관상동맥질환", "협심증",
    "심근경색", "심부전", "암",
    "심장질환_허혈성또는심부전",
    "고혈압", "고지혈증"
]

missing = [
    c for c in BASIC_COLS + DIAGNOSIS_COLS
    if c not in health_data.columns
]

if missing:
    raise ValueError(f"원본에 없는 컬럼: {missing}")

HEALTH_COLS = [
    c for c in health_data.columns
    if c not in BASIC_COLS + DIAGNOSIS_COLS
]

OUTPUT_COLS = [
    "participant", "성별", "나이", "질병",
    "진단여부", "위험군분류", "근거", "분석한계"
]

print("참가자 수:", len(health_data))
print("건강검진 열 수:", len(HEALTH_COLS))
````

````python
# ========================================
# 3. 진단여부 계산
# ========================================

def get_diagnosis(row, disease):
    cols = DIAGNOSIS_MAP[disease]

    if not cols:
        return None

    values = pd.to_numeric(row[cols], errors="coerce")

    if (values == 1).any():
        return 1

    if values.isna().any():
        return None

    if (values == 0).all():
        return 0

    return None
````

````python
# ========================================
# 4. 데이터 JSON 변환
# ========================================

def clean_value(value):
    if isinstance(value, (list, dict)):
        return value

    if pd.isna(value):
        return None

    if isinstance(value, np.integer):
        return int(value)

    if isinstance(value, np.floating):
        return float(value)

    if isinstance(value, np.bool_):
        return bool(value)

    return value


def make_input(row):
    return {
        "participant": clean_value(row["participant"]),
        "성별": clean_value(row["성별"]),
        "나이": clean_value(row["나이"]),
        "기존진단이력": {
            c: clean_value(row[c])
            for c in DIAGNOSIS_COLS
        },
        "건강검진정보": {
            c: clean_value(row[c])
            for c in HEALTH_COLS
        }
    }


# ========================================
# 5. Claude 시스템 프롬프트
# ========================================

SYSTEM_PROMPT = """
당신은 건강검진 데이터의 위험 신호를 정리하는
의학 연구 보조 AI입니다.

참가자의 기존 진단 이력과 건강검진 수치를 이용해
다음 6개 질병에 대한 프로젝트용 위험군을 분류하세요.

대상 질병:
당뇨, 심장, 뇌혈관, 신장, 간, 고혈압

[위험군 정의]

해당없음:
질병이 적용되지 않는다는 명확한 근거가 있는 경우.
정상 검사 결과나 성별만으로 해당없음 처리하지 마세요.

일반군:
제공된 정보에서 해당 질병과 관련된
뚜렷한 이상 신호가 확인되지 않는 경우.

위험군:
관련 검사 수치 이상 또는 위험요인이 있어
추가적인 평가가 필요한 경우.

고위험군:
현저한 검사 수치 이상이나 여러 위험요인의 중첩으로
우선적인 의학적 평가가 필요한 경우.

평가불가:
중요한 검사 정보가 너무 부족하거나
근거가 불충분한 경우.

[질병별 평가 요소]

당뇨:
공복혈당, HbA1c, BMI, 기존 당뇨 진단

심장:
총콜레스테롤, LDL, HDL, 중성지방,
혈압, 당뇨, 고지혈증, BMI, 나이,
기존 심장질환 진단

뇌혈관:
혈압, 당뇨, 이상지질혈증,
BMI, 나이, 기존 뇌졸중 진단

신장:
크레아티닌, 당뇨, 고혈압, 나이
eGFR 및 소변 알부민이 없으면
만성콩팥병을 확진하지 마세요.

간:
AST, ALT, GGT, BMI, 중성지방, 당뇨
간효소 수치만으로 간경변이나
간섬유화를 확진하지 마세요.

고혈압:
수축기혈압, 이완기혈압,
BMI, 나이, 기존 고혈압 진단

[분석 규칙]

1. 제공된 데이터만 사용하세요.
2. 결측치를 정상으로 가정하지 마세요.
3. 기존 진단 여부와 향후 발병 위험을 구분하세요.
4. 진단이 있다고 자동으로 고위험군으로 분류하지 마세요.
5. 검사 수치와 단위를 확인하세요.
6. 단일 검사만으로 질병을 확진하지 마세요.
7. 위험기준_초과항목은 참고자료로만 사용하세요.
8. 근거에는 실제 검사 수치와 위험요인을 기재하세요.
9. 분석한계에는 누락 정보 및 추가 검사 필요성을 설명하세요.
10. 근거가 부족하면 평가불가로 분류하지만, 너무 많이 평가불가로 분류하지 마세요. 최대한 분석 근거를 찾아서 위험군을 분류하세요.
11. 위험군은 검증된 발병확률이 아니라
    프로젝트용 건강검진 선별 등급입니다.
12. 반드시 6개 질병을 각각 한 번씩 반환하세요.

[질병명 출력 규칙 — 매우 중요]

질병 필드에는 반드시 다음 6개 문자열만 사용하세요.
1. 당뇨
2. 심장
3. 뇌혈관
4. 신장
5. 간
6. 고혈압

질병명은 절대로 수정하거나 확장하지 마세요.

[출력 길이 제한]

1. 각 질병의 근거는 1~2문장, 120자 이내로 작성하세요.
2. 각 질병의 분석한계는 1문장, 80자 이내로 작성하세요.
3. 불필요한 설명이나 반복을 생략하세요.
4. 반드시 6개 질병의 결과를 모두 포함하세요.
5. JSON 이외의 텍스트는 출력하지 마세요.

반환 JSON 형식:
{
  "results": [
    {
      "질병": "당뇨",
      "위험군분류": "일반군",
      "근거": "구체적인 판단 근거",
      "분석한계": "분석의 한계"
    }
  ]
}

results 배열에는 반드시 6개 질병을 각각 정확히 한 번씩 포함하세요.
"""
````

````python
# ========================================
# 6. Claude 호출 함수
# ========================================

VALID_GROUPS = {
    "해당없음", "일반군", "위험군",
    "고위험군", "평가불가"
}

def classify_participant(row):
    payload = make_input(row)

    response = client.messages.create(
        model=MODEL,
        max_tokens=8000,
        system=SYSTEM_PROMPT,
        messages=[
            {
                "role": "user",
                "content": (
                    "다음 참가자의 건강검진 정보를 분석하고 "
                    "6개 질병을 분류하세요.\n\n"
                    + json.dumps(payload, ensure_ascii=False)
                )
            }
        ]
    )

    if response.stop_reason != "end_turn":
        raise RuntimeError(
            f"응답이 정상 완료되지 않음: {response.stop_reason}"
        )

    answer = "".join(
        block.text
        for block in response.content
        if block.type == "text"
    ).strip()

    # Markdown 코드펜스가 포함된 경우 제거
    if answer.startswith("```"):
        answer = answer.split("\n", 1)[1]
        answer = answer.rsplit("```", 1)[0].strip()

    parsed = json.loads(answer)
    results = parsed["results"]

    names = [item["질병"] for item in results]

    if len(names) != 6 or set(names) != set(DISEASES):
        raise ValueError(f"질병 결과 불일치: {names}")

    output = []

    for item in results:
        disease = item["질병"]
        group = item["위험군분류"]

        if group not in VALID_GROUPS:
            raise ValueError(f"잘못된 위험군: {group}")

        output.append({
            "participant": row["participant"],
            "성별": row["성별"],
            "나이": row["나이"],
            "질병": disease,
            "진단여부": get_diagnosis(row, disease),
            "위험군분류": group,
            "근거": item["근거"],
            "분석한계": item["분석한계"]
        })

    return output
````

````python
# ========================================
# 7. 참가자 1명 테스트
# ========================================

test_result = classify_participant(health_data.iloc[0])
test_df = pd.DataFrame(test_result, columns=OUTPUT_COLS)

display(test_df)
````

````python
all_results = []
failed_participants = []

for i, (_, row) in enumerate(health_data.iterrows(), start=1):

    participant = row["participant"]

    print(f"[{i}/{len(health_data)}] {participant} 분석 중...")

    try:
        # 참가자 1명 → 질병 6개 분석
        result = classify_participant(row)

        # Long Form 데이터 누적
        all_results.extend(result)

        print(f"완료! 누적 {len(all_results)}행")

    except Exception as e:
        failed_participants.append({
            "participant": participant,
            "error": str(e)
        })

        print(f"오류 발생: {e}")

    # 매 참가자마다 중간 저장
    pd.DataFrame(
        all_results,
        columns=OUTPUT_COLS
    ).to_excel(
        "health_risk_long_partial.xlsx",
        index=False
    )

    time.sleep(0.5)


# 최종 DataFrame 생성
health_risk_long = pd.DataFrame(
    all_results,
    columns=OUTPUT_COLS
)

# 최종 엑셀 저장
health_risk_long.to_excel(
    "health_risk_long.xlsx",
    index=False
)

# 실패한 참가자 저장
failed_df = pd.DataFrame(failed_participants)
failed_df.to_excel(
    "health_risk_failed.xlsx",
    index=False
)

print("\n===== 분석 완료 =====")
print("전체 참가자:", len(health_data))
print("분석 성공:", health_risk_long["participant"].nunique(), "명")
print("분석 실패:", len(failed_participants), "명")
print("최종 데이터:", health_risk_long.shape)

display(health_risk_long.head(12))
````

````python
# 참가자별 질병 행 수 확인
print(health_risk_long.groupby("participant")["질병"].nunique().value_counts())

# 위험군 분포
print(health_risk_long["위험군분류"].value_counts())

# 질병별 위험군 분포
display(
    pd.crosstab(
        health_risk_long["질병"],
        health_risk_long["위험군분류"]
    )
)
````

## DNA 결과 (연구소데이터) 위험군 분류

원래 연구소가 해주는 데이터라서, 이 작업은 LLM에게 시킬 것임

````python
import os
import re
import json
import time
import numpy as np
import pandas as pd

from dotenv import load_dotenv
from anthropic import Anthropic

load_dotenv(override=True)

API_KEY = os.getenv("CLAUDE_API_KEY")
MODEL = os.getenv("CLAUDE_DEFAULT_MODEL", "claude-haiku-4-5")

if not API_KEY:
    raise RuntimeError("CLAUDE_API_KEY를 확인하세요.")

client = Anthropic(
    api_key=API_KEY,
    timeout=180.0,
    max_retries=2
)

dna_data = pd.read_excel("test_DNA_data.xlsx")

DISEASES = [
    "당뇨", "심장", "뇌혈관",
    "유방암", "대장암", "치매"
]

META_COLS = [
    "participant", "성별", "나이",
    "DNA_전체SNP수", "DNA_미판독수",
    "DNA_패널SNP확보수"
]

SNP_COLS = [
    col for col in dna_data.columns
    if col not in META_COLS
]

OUTPUT_COLS = [
    "participant", "성별", "나이",
    "질병", "위험군분류", "근거", "분석한계"
]

print("모델:", MODEL)
print("참가자:", len(dna_data))
print("SNP 열:", len(SNP_COLS))
print("예상 결과:", len(dna_data) * 6, "행")
````

````python
SYSTEM_PROMPT = """
당신은 유전학과 질병 위험요인 분석을 지원하는
AI 유전체 연구 보조 시스템입니다.

[목표]

참가자의 SNP 유전자형을 분석하여 다음 6개 질병의
유전적 위험 신호를 평가하고, 프로젝트 시연용
상대적 위험군을 분류하세요.

대상 질병:
당뇨, 심장, 뇌혈관, 유방암, 대장암, 치매

이번 작업은 교육용 데이터 분석 프로젝트입니다.
의학적 진단이나 실제 발병확률 예측이 아닙니다.

[분석 방식]

1. 제공된 SNP 중 각 질병과 관련성이 있는 변이를
   당신이 알고 있는 유전학적 지식을 바탕으로 선별하세요.

2. 모든 SNP를 동일하게 취급하지 말고
   질병과의 관련성, 연구 근거, 알려진 효과의 중요도를
   고려하세요.

3. SNP 컬럼명에 질병명이 포함되어 있으면
   해당 질병의 분석 후보로 고려하되,
   그것만으로 위험 변이라고 판단하지 마세요.

4. 특정 SNP의 위험 대립유전자나 효과 방향을
   확실히 알지 못하면 위험도 계산에 사용하지 마세요.

5. 참가자의 실제 유전자형과 알려진 위험 대립유전자를
   비교하여 위험 신호를 설명하세요.

6. 여러 위험 SNP가 존재하더라도
   검증된 가중치가 없으면 위험도를 단순 합산하거나
   정확한 발병확률을 만들어내지 마세요.

7. 결측 SNP는 정상 유전자형으로 취급하지 마세요.

8. II, DD 등의 삽입·결실 유전자형은
   변이 방향이 확인되지 않으면
   병원성 변이 보유 여부를 확정하지 마세요.

9. 참가자의 성별과 나이는 참고 정보로만 사용하고,
   검증되지 않은 가중치를 임의로 부여하지 마세요.

[질병별 핵심 분석]

당뇨:
제2형 당뇨병의 유전적 소인을 중심으로 분석합니다.
TCF7L2, FTO 및 당뇨 관련 SNP를 검토하세요.

심장:
관상동맥질환 중심의 유전적 소인을 분석합니다.
9p21, LPA 및 심장 관련 SNP를 검토하세요.
심부전, 부정맥, 관상동맥질환의 차이를 고려하세요.

뇌혈관:
뇌졸중 관련 유전적 소인을 분석합니다.
PITX2, HDAC9 및 뇌졸중 관련 SNP를 검토하세요.
허혈성 뇌졸중과 출혈성 뇌졸중을 혼동하지 마세요.

유방암:
FGFR2, BRCA1, BRCA2 및 유방암 관련 SNP를 검토하세요.
일반적인 GWAS 위험 변이와 병원성 변이를 구분하세요.
BRCA 유전자형의 II/DD만으로 병원성 여부를
확정하지 마세요.
남성도 유방암이 발생할 수 있습니다.

대장암:
8q24, SMAD7 및 대장암 관련 SNP를 검토하세요.

치매:
APOE rs429358, rs7412를 우선 검토하세요.
두 SNP가 모두 존재하고 대립유전자 방향을
해석할 수 있는 경우에만 APOE 유형을 설명하세요.
알츠하이머병 유전적 소인과 전체 치매를 구분하세요.

[위험군 분류 기준]

위험군은 다음 네 가지 중 하나로 분류하세요.

1. 해당없음
질병 자체가 적용되지 않는다는 명확한 근거가
있는 경우에만 사용합니다.
성별이나 결측을 이유로 해당없음 처리하지 마세요.

2. 일반군
제공된 SNP 중 확실하게 해석 가능한 정보에서
뚜렷한 위험 증가 신호가 확인되지 않는 경우입니다.

3. 위험군
해석 가능한 SNP에서 해당 질병과 관련된
유전적 위험 증가 신호가 확인되는 경우입니다.

4. 고위험군
해당 질병과의 관련성이 강하고 잘 알려진
유전적 위험 신호가 확인되며,
일반적인 위험군보다 주목할 근거가 있는 경우입니다.
단순히 위험 SNP의 개수가 많다는 이유만으로
고위험군으로 분류하지 마세요.

[결측치 처리]

- 결측 SNP는 분석에서 제외하세요.
- SNP 데이터가 부족해도 시연용 분류를 수행하세요.
- 정보가 부족한 경우 일반군으로 잠정 분류할 수 있지만,
  이는 유전적 위험이 낮다는 의미가 아닙니다.
- 결측이 많거나 해석 가능한 SNP가 적으면
  반드시 분석한계에 불확실성을 설명하세요.
- 위험군을 억지로 높이지 마세요.
- 검증되지 않은 위험 대립유전자나 효과 크기를
  임의로 생성하지 마세요.

[근거 작성]

- 각 질병별로 주요 SNP 1~4개를 우선 설명하세요.
- 실제 참가자의 SNP 유전자형을 명시하세요.
- 위험 증가 신호인지, 근거가 제한적인지 구분하세요.
- 해석 가능한 관련 SNP가 없으면 그 사실을 설명하세요.
- 근거는 150자 이내로 작성하세요.

[분석한계 작성]

- 결측 SNP
- 제한된 유전자 패널
- 인구집단별 차이
- 검증된 PRS 모델 부재
- 실제 발병확률 예측 불가

위 항목 중 해당하는 내용을 100자 이내로 작성하세요.

[출력 규칙]

반드시 다음 6개 질병을 정확히 한 번씩 반환하세요.

당뇨, 심장, 뇌혈관, 유방암, 대장암, 치매

질병명은 절대 변경하지 마세요.
위험군분류는 반드시 다음 네 가지 중 하나입니다.

해당없음 / 일반군 / 위험군 / 고위험군

반환 형식:

{
  "results": [
    {
      "질병": "당뇨",
      "위험군분류": "일반군",
      "근거": "실제 SNP 유전자형과 관련성 설명",
      "분석한계": "결측치와 해석의 한계"
    }
  ]
}

반드시 6개 결과를 반환하세요.
JSON 외의 텍스트는 출력하지 마세요.
"""
````

````python
def clean_value(value):
    if pd.isna(value):
        return None

    if isinstance(value, np.integer):
        return int(value)

    if isinstance(value, np.floating):
        return float(value)

    return value


def make_dna_input(row):

    snps = {}

    missing_count = 0

    for col in SNP_COLS:
        value = clean_value(row[col])

        if value is None or str(value).strip() in [
            "", "--", "NA", "nan"
        ]:
            missing_count += 1
            continue

        snps[col] = str(value).strip()

    return {
        "participant": clean_value(row["participant"]),
        "성별": clean_value(row["성별"]),
        "나이": clean_value(row["나이"]),
        "전체패널SNP수": len(SNP_COLS),
        "확보SNP수": len(snps),
        "결측SNP수": missing_count,
        "SNP유전자형": snps
    }
````

````python
VALID_GROUPS = {
    "해당없음", "일반군", "위험군", "고위험군"
}

DISEASE_NAME_MAP = {
    "당뇨병": "당뇨",
    "심장질환": "심장",
    "뇌졸중": "뇌혈관",
    "뇌혈관질환": "뇌혈관",
    "유방암": "유방암",
    "대장암": "대장암",
    "알츠하이머": "치매",
    "알츠하이머병": "치매"
}


def classify_dna_participant(row):

    payload = make_dna_input(row)

    response = client.messages.create(
        model=MODEL,
        max_tokens=8000,
        system=SYSTEM_PROMPT,
        messages=[
            {
                "role": "user",
                "content": (
                    "아래 참가자의 SNP 유전자형을 분석하여 "
                    "6개 질병의 시연용 위험군을 분류하세요.\n\n"
                    + json.dumps(
                        payload,
                        ensure_ascii=False
                    )
                )
            }
        ]
    )

    if response.stop_reason != "end_turn":
        raise RuntimeError(
            f"응답이 중단됨: {response.stop_reason}"
        )

    answer = "".join(
        block.text
        for block in response.content
        if block.type == "text"
    ).strip()

    if answer.startswith("```"):
        answer = answer.split("\n", 1)[1]
        answer = answer.rsplit("```", 1)[0].strip()

    parsed = json.loads(answer)
    results = parsed["results"]

    # 질병명 보정
    for item in results:
        disease = item["질병"].strip()
        item["질병"] = DISEASE_NAME_MAP.get(
            disease, disease
        )

    names = [item["질병"] for item in results]

    if len(names) != 6 or set(names) != set(DISEASES):
        raise ValueError(
            f"질병 결과 불일치: {names}"
        )

    long_rows = []

    for item in results:

        disease = item["질병"]
        group = item["위험군분류"]

        if group not in VALID_GROUPS:
            raise ValueError(
                f"잘못된 위험군: {group}"
            )

        long_rows.append({
            "participant": row["participant"],
            "성별": row["성별"],
            "나이": row["나이"],
            "질병": disease,
            "위험군분류": group,
            "근거": item["근거"],
            "분석한계": item["분석한계"]
        })

    return long_rows
````

````python
test_result = classify_dna_participant(
    dna_data.iloc[0]
)

test_df = pd.DataFrame(
    test_result,
    columns=OUTPUT_COLS
)

display(test_df)
````

````python
PARTIAL_FILE = "dna_risk_long_partial.xlsx"
FINAL_FILE = "dna_risk_long.xlsx"
FAILED_FILE = "dna_risk_failed.xlsx"

# 기존 성공 결과 불러오기
if os.path.exists(PARTIAL_FILE):
    existing_df = pd.read_excel(PARTIAL_FILE)

    all_results = existing_df.to_dict("records")

    completed_ids = set(
        existing_df.groupby("participant")["질병"]
        .nunique()
        .loc[lambda x: x == 6]
        .index
    )

    # 불완전한 참가자 결과는 제거
    all_results = [
        r for r in all_results
        if r["participant"] in completed_ids
    ]

else:
    all_results = []
    completed_ids = set()

failed_participants = []

for i, (_, row) in enumerate(dna_data.iterrows(), start=1):

    participant = row["participant"]

    if participant in completed_ids:
        print(f"[{i}/{len(dna_data)}] {participant} 건너뜀")
        continue

    print(f"[{i}/{len(dna_data)}] {participant} 분석 중")

    try:
        result = classify_dna_participant(row)

        all_results.extend(result)
        completed_ids.add(participant)

        print("완료")

    except Exception as e:
        failed_participants.append({
            "participant": participant,
            "error": str(e)
        })

        print("오류:", e)

    # 중간 저장
    pd.DataFrame(
        all_results,
        columns=OUTPUT_COLS
    ).to_excel(
        PARTIAL_FILE,
        index=False
    )

    time.sleep(0.5)


# 최종 데이터
dna_risk_long = pd.DataFrame(
    all_results,
    columns=OUTPUT_COLS
)

dna_risk_long.to_excel(
    FINAL_FILE,
    index=False
)

pd.DataFrame(
    failed_participants,
    columns=["participant", "error"]
).to_excel(
    FAILED_FILE,
    index=False
)

print("\n===== 분석 완료 =====")
print("성공 참가자:", dna_risk_long["participant"].nunique())
print("실패 참가자:", len(failed_participants))
print("최종 데이터:", dna_risk_long.shape)

display(dna_risk_long.head(12))
````
