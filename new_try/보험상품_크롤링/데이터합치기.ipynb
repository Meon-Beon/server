```python
import pandas as pd
```

## Longform 데이터 합치기

```python
# CSV 파일 불러오기
df_상해 = pd.read_csv("상해_product_longform.csv", encoding="utf-8-sig")
df_암 = pd.read_csv("암_product_longform.csv", encoding="utf-8-sig")
df_질병 = pd.read_csv("질병_product_longform.csv", encoding="utf-8-sig")
df_치매 = pd.read_csv("치매_product_longform.csv", encoding="utf-8-sig")
```

```python
df_상해 = df_상해.drop(columns=["기간"])
df_질병 = df_질병.drop(columns=["기간"])

df_상해["보험종류"] = "상해"
df_암["보험종류"] = "암"
df_질병["보험종류"] = "질병"
df_치매["보험종류"] = "치매"
```

```python
보험상품_longform = pd.concat(
    [df_상해, df_암, df_질병, df_치매],
    axis=0,
    ignore_index=True
)
```

```python
display(보험상품_longform.head())
print(보험상품_longform.shape)
```

```python
보험상품_longform.to_csv("보험상품_longform.csv", index=False, encoding="utf-8-sig")
```

## Wideform 데이터 합치기

```python
# CSV 파일 불러오기
df_상해 = pd.read_csv("상해_product_wideform.csv", encoding="utf-8-sig")
df_암 = pd.read_csv("암_product_wideform.csv", encoding="utf-8-sig")
df_질병 = pd.read_csv("질병_product_wideform.csv", encoding="utf-8-sig")
df_치매 = pd.read_csv("치매_product_wideform.csv", encoding="utf-8-sig")
```

```python
df_상해 = df_상해.drop(columns=["기간"])
df_암 = df_암.drop(columns=["번호"])
df_질병 = df_질병.drop(columns=["기간"])
df_치매 = df_치매.drop(columns=["순번"])

df_상해["보험종류"] = "상해"
df_암["보험종류"] = "암"
df_질병["보험종류"] = "질병"
df_치매["보험종류"] = "치매"
```

```python
보험상품_wideform = pd.concat(
    [df_상해, df_암, df_질병, df_치매],
    axis=0,
    ignore_index=True
)
```

```python
보험상품_wideform.to_csv("보험상품_wideform.csv", index=False, encoding="utf-8-sig")
```
